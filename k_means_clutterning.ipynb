```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.cluster import KMeans
sns.set()
```

```python
#load the data
data=pd.read_csv("C:\\Users\\user\\Downloads\\archive\\3.01.Countryclusters.csv")
data
```

```python
#plotting the dataset
plt.scatter(data['Longitude'],data['Latitude'])
plt.xlim(-180,180)
plt.ylim(-90,90)
plt.show()
```

```python
#selecting the feature
x=data.iloc[:,1:3]
x
```

```python
#clustering
kmeans=KMeans(5)
kmeans.fit(x)
#clustering result
#fitting the data into the model
identified_clusters=kmeans.fit_predict(x)
#the identified clusters where our data can go identified_clusters
#creating a copy of our data
data_with_clusters=data.copy()
#new clusters--->data
data_with_clusters['Clusters']=identified_clusters
#plot the data
plt.scatter(data_with_clusters['Longitude'],data_with_clusters['Latitude'],c=data_with_clusters['Clusters'],cmap='rainbow')
```

```python
# import packages
import math
```

```python
# Dataset
data = [
    (10, 2),
    (12, 4),
    (16, 6),
    (20, 8),
    (24, 10),
    (26, 12),
    (30, 14),
    (34, 16)
]

# Initial centroids
c1 = (10, 2)
c2 = (34, 16)

# K-Means for K = 2
for iteration in range(5):

    cluster1 = []
    cluster2 = []

    # Assign points to nearest centroid
    for p in data:

        d1 = math.sqrt(
            (p[0] - c1[0])**2 +
            (p[1] - c1[1])**2
        )

        d2 = math.sqrt(
            (p[0] - c2[0])**2 +
            (p[1] - c2[1])**2
        )

        if d1 < d2:
            cluster1.append(p)
        else:
            cluster2.append(p)

    # Calculate new centroid for Cluster 1
    c1 = (
        sum(p[0] for p in cluster1) / len(cluster1),
        sum(p[1] for p in cluster1) / len(cluster1)
    )

    # Calculate new centroid for Cluster 2
    c2 = (
        sum(p[0] for p in cluster2) / len(cluster2),
        sum(p[1] for p in cluster2) / len(cluster2)
    )

    print("Iteration:", iteration + 1)
    print("Cluster 1:", cluster1)
    print("Cluster 2:", cluster2)
    print("Centroid 1:", c1)
    print("Centroid 2:", c2)
    print()

# Convert to DataFrame
df1 = pd.DataFrame(cluster1, columns=["X", "Y"])
df2 = pd.DataFrame(cluster2, columns=["X", "Y"])

# Scatter plot
plt.scatter(df1["X"], df1["Y"], label="Cluster 1", s=100)
plt.scatter(df2["X"], df2["Y"], label="Cluster 2", s=100)

# Plot centroids
plt.scatter(c1[0], c1[1], marker="X", s=250, label="Centroid 1")
plt.scatter(c2[0], c2[1], marker="X", s=250, label="Centroid 2")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("K-Means Clustering (K=2)")
plt.legend()
plt.show()
```
